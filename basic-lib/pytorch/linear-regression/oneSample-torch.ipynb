```python
import numpy as np 
import torch
import torch.nn as nn

data = np.genfromtxt("data.csv", delimiter=',') 
x_data = torch.from_numpy(data[:, 0:1])
y_data = torch.from_numpy(data[:, 1])

print('x', x_data)
print('y', y_data)
```

```python
linear = nn.Linear(1, 1) # One linear layer
linear.weight.data = torch.tensor([[-0.34]]) # Set weight
linear.bias.data = torch.Tensor([0.04]) # Set bias

print(linear.weight)
print(linear.bias)
```

```python
loss_fn = torch.nn.MSELoss() # Choose loss
optimizer = torch.optim.SGD(linear.parameters(), lr=0.01) # Choose optim
epochs = 1
```

```python
for epoch in range(epochs):
    for i in range(4):
        x = x_data[i].float()
        y = y_data[i:i+1].float()

        # y hat
        y_hat = linear(x)
        print(f"y_hat: {y_hat.item()}")

        # Loss
        loss = loss_fn(y_hat, y)
        print(f"Loss: {loss.item()}")

        # Compute gradient
        loss.backward()
        print(f"(db, dw): ({linear.bias.grad.data}, {linear.weight.grad.data})")

        # Update
        optimizer.step()
        optimizer.zero_grad()
        print(f"(b, w): ({linear.bias.data}, {linear.weight.data})\n\n")
```
